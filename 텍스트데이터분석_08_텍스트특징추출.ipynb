# 텍스트 특징 추출(TF-IDF)

### 1. 샘플 코퍼스로 연습

```python
sample_corpus = [
'자연어처리 강의를 시작하겠습니다.',
'자연어처리는 재미있습니다.',
'밥을 먹고 강의를 듣고 있습니다.',
'이번 자연어처리 강의는 한국어 자연어처리입니다.'
]
```

```python
import sklearn
```

```python
from konlpy.tag import Okt
def my_tokenizer(text):
    return Okt().nouns(text)
```

```python
# TfidfVectorizer 객체 생성
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(tokenizer=my_tokenizer)
# 특징 집합과 관련 데이터 모델을 생성
vectorizer.fit(sample_corpus)
print(vectorizer.get_feature_names_out())
# 특징 벡터를 추출
sample_dtm = vectorizer.transform(sample_corpus)
# type(sample_dtm)
print(sample_dtm.toarray())
```

## 2. 다음 영화 리뷰로 연습

```python
#max_feature = 10000
import pandas as pd
datafile = './data/daum_movie_review.csv'
data_df = pd.read_csv(datafile)
```

```python
reviews = list(data_df.review)
```

```python
from konlpy.tag import Okt
def my_tokenizer(text):
    return Okt().nouns(text)
```

```python
import sklearn
# TfidfVectorizer 객체 생성
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(tokenizer=my_tokenizer, max_features=100)
# 특징 집합과 관련 데이터 모델을 생성
vectorizer.fit(reviews)
print(vectorizer.get_feature_names_out())
# 특징 벡터를 추출
reviews_dtm = vectorizer.transform(reviews)
# type(sample_dtm)
print(reviews_dtm.toarray())
```
